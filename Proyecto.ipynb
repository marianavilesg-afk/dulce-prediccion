```python
import pandas as pd

# Crear la base de datos de ventas
datos = {
    "Fecha": [
        "2026-09-01", "2026-09-01",
        "2026-09-02", "2026-09-02",
        "2026-09-03", "2026-09-03",
        "2026-09-04", "2026-09-04",
        "2026-09-05", "2026-09-05",
        "2026-09-06", "2026-09-06",
        "2026-09-07", "2026-09-07",
        "2026-09-08", "2026-09-08",
        "2026-09-09", "2026-09-09",
        "2026-09-10", "2026-09-10"
    ],

    "Postre": [
        "Brownie", "Galleta",
        "Dona", "Porción de torta",
        "Brownie", "Galleta",
        "Dona", "Porción de torta",
        "Brownie", "Galleta",
        "Dona", "Porción de torta",
        "Brownie", "Galleta",
        "Dona", "Porción de torta",
        "Brownie", "Galleta",
        "Dona", "Porción de torta"
    ],

    "Dia": [
        "Martes", "Martes",
        "Miércoles", "Miércoles",
        "Jueves", "Jueves",
        "Viernes", "Viernes",
        "Sábado", "Sábado",
        "Domingo", "Domingo",
        "Lunes", "Lunes",
        "Martes", "Martes",
        "Miércoles", "Miércoles",
        "Jueves", "Jueves"
    ],

    "Precio": [
        5000, 2500, 3500, 7000,
        5000, 2500, 3500, 7000,
        5000, 2500, 3500, 7000,
        5000, 2500, 3500, 7000,
        5000, 2500, 3500, 7000
    ],

    "Promocion": [
        "No", "Sí", "No", "Sí",
        "Sí", "No", "Sí", "No",
        "No", "Sí", "No", "Sí",
        "Sí", "No", "Sí", "No",
        "No", "Sí", "No", "Sí"
    ],

    "Preparadas": [
        30, 40, 25, 20,
        35, 30, 35, 18,
        40, 45, 30, 25,
        32, 35, 28, 22,
        30, 40, 25, 24
    ],

    "Vendidas": [
        24, 36, 20, 18,
        31, 22, 32, 13,
        35, 42, 27, 23,
        29, 28, 25, 16,
        25, 37, 19, 22
    ]
}

df = pd.DataFrame(datos)

# Mostrar la base de datos
df
```

```python
# Seleccionar las variables que utilizaremos
X = df[["Postre", "Dia", "Precio", "Promocion", "Preparadas"]]

# Seleccionar lo que queremos predecir
y = df["Vendidas"]

# Convertir las palabras en valores numéricos
X = pd.get_dummies(X, columns=["Postre", "Dia", "Promocion"], dtype=int)

# Mostrar las variables preparadas
X.head()
```

```python
from sklearn.model_selection import train_test_split

# Separar los datos en entrenamiento y prueba
X_entrenamiento, X_prueba, y_entrenamiento, y_prueba = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

# Mostrar la cantidad de datos
print("Datos de entrenamiento:", len(X_entrenamiento))
print("Datos de prueba:", len(X_prueba))
```

```python
from sklearn.tree import DecisionTreeRegressor

modelo = DecisionTreeRegressor(max_depth=3, random_state=42)

modelo.fit(X_entrenamiento, y_entrenamiento)

print("¡Modelo entrenado correctamente!")
```

```python
y_predicciones = modelo.predict(X_prueba)

print("Ventas reales:", list(y_prueba))
print("Ventas predichas:", list(y_predicciones))
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_prueba, y_predicciones)
rmse = np.sqrt(mean_squared_error(y_prueba, y_predicciones))
r2 = r2_score(y_prueba, y_predicciones)

print("Error absoluto medio (MAE):", round(mae, 2))
print("Raíz del error cuadrático medio (RMSE):", round(rmse, 2))
print("Coeficiente de determinación (R²):", round(r2, 2))
```

```python
comparacion = pd.DataFrame({
    "Ventas reales": list(y_prueba),
    "Ventas predichas": np.round(y_predicciones, 2)
})

comparacion["Error"] = np.round(
    comparacion["Ventas reales"] - comparacion["Ventas predichas"], 2
)

display(comparacion)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Campos que podrá completar el usuario
postre = widgets.Dropdown(
    options=["Brownie", "Dona", "Galleta", "Porción de torta"],
    description="Postre:"
)

dia = widgets.Dropdown(
    options=["Lunes", "Martes", "Miércoles", "Jueves",
             "Viernes", "Sábado", "Domingo"],
    description="Día:"
)

precio = widgets.IntText(
    value=5000,
    description="Precio ($):"
)

promocion = widgets.Dropdown(
    options=["No", "Sí"],
    description="Promoción:"
)

preparadas = widgets.IntText(
    value=30,
    description="Preparadas:"
)

display(postre, dia, precio, promocion, preparadas)
```

```python
# Botón para generar la predicción
boton_predecir = widgets.Button(
    description="Predecir ventas",
    button_style="success",
    icon="chart-line"
)

# Espacio para mostrar el resultado
resultado = widgets.Output()

def predecir_ventas(b):
    with resultado:
        clear_output()

        # Validar los datos ingresados
        if precio.value <= 0 or preparadas.value < 0:
            print("Por favor, ingresa un precio positivo y una cantidad preparada válida.")
            return

        # Crear una fila con las mismas columnas del entrenamiento
        entrada = pd.DataFrame(0, index=[0], columns=X.columns)

        entrada["Precio"] = precio.value
        entrada["Preparadas"] = preparadas.value
        entrada[f"Postre_{postre.value}"] = 1
        entrada[f"Dia_{dia.value}"] = 1
        entrada[f"Promocion_{promocion.value}"] = 1

        # Obtener la predicción del modelo
        prediccion = modelo.predict(entrada)[0]

        # Mostrar el resultado
        print("🍰 PREDICCIÓN DE DEMANDA")
        print("------------------------------")
        print(f"Postre seleccionado: {postre.value}")
        print(f"Día: {dia.value}")
        print(f"Unidades preparadas: {preparadas.value}")
        print(f"Ventas estimadas: {max(0, round(prediccion))} unidades")

# Conectar el botón con la función
boton_predecir.on_click(predecir_ventas)

display(boton_predecir, resultado)
```
